```python
import matplotlib.pyplot as plt

type_counts = df['type'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(type_counts.index, type_counts.values)

plt.title('Netflix Movies vs TV Shows', fontsize=16)
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')

plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()

plt.savefig('movie_vs_tv_shows.png', dpi=300)
plt.show()
```

```python
year_counts = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 5))

plt.plot(year_counts.index, year_counts.values)

plt.title('Netflix Titles by Release Year', fontsize=16)
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()

plt.savefig('titles_by_release_year.png', dpi=300)
plt.show()
```

```python
type_counts = df['type'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Netflix Movies vs TV Shows', fontsize=16)

plt.tight_layout()

plt.savefig('movie_tv_pie_chart.png', dpi=300)
plt.show()
```

```python
country_counts = (
    df['country']
    .dropna()
    .str.split(', ')
    .explode()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10, 6))

plt.barh(country_counts.index, country_counts.values)

plt.title('Top 10 Countries by Netflix Titles', fontsize=16)
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.gca().invert_yaxis()
plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.tight_layout()

plt.savefig('top_10_countries.png', dpi=300)
plt.show()

```

```python
rating_counts = df['rating'].value_counts().head(10)

plt.figure(figsize=(10, 6))

plt.bar(rating_counts.index, rating_counts.values)

plt.title('Top 10 Netflix Content Ratings', fontsize=16)
plt.xlabel('Rating')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.tight_layout()

plt.savefig('top_10_ratings.png', dpi=300)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

year_type = df.groupby(['release_year', 'type']).size().reset_index(name='count')

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=year_type,
    x='release_year',
    y='count',
    hue='type'
)

plt.title('Movies vs TV Shows by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.show()
```
